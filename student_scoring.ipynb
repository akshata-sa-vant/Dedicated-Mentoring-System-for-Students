```python
import pandas as pd

df = pd.read_csv("Students.csv")

# ---------- APS ----------
df["gpa_score"] = (df["gpa"] / 10) * 100
df["APS"] = (
    0.5 * df["gpa_score"] +
    0.3 * df["attendance"] +
    0.2 * df["assignments_completion"]
)

# ---------- WWS ----------
df["stress_score"] = (10 - df["stress_level"]) * 10

def sleep_score(hours):
    if 7 <= hours <= 9:
        return 100
    elif hours == 6 or hours == 10:
        return 80
    else:
        return 60

df["sleep_score"] = df["sleep_hours"].apply(sleep_score)
df["mental_score"] = df["mental_wellbeing"] * 10

df["WWS"] = (
    df["stress_score"] +
    df["sleep_score"] +
    df["mental_score"]
) / 3

# ---------- PTMS ----------
df["productivity_scaled"] = df["productivity_score"] * 10
df["distraction_score"] = (10 - df["distractions"]) * 10

df["PTMS"] = (
    df["productivity_scaled"] +
    df["distraction_score"] +
    df["engagement_score"]
) / 3

# ---------- CRS ----------
df["career_score"] = df["career_clarity"] * 10
df["skill_score"] = df["skill_readiness"] * 10

df["CRS"] = (df["career_score"] + df["skill_score"]) / 2

# ---------- SRI ----------
df["SRI"] = (
    0.3 * df["APS"] +
    0.2 * df["WWS"] +
    0.2 * df["PTMS"] +
    0.3 * df["CRS"]
)

# ---------- Classification ----------
def classify(sri):
    if sri >= 80:
        return "Green"
    elif sri >= 65:
        return "Blue"
    elif sri >= 50:
        return "Yellow"
    else:
        return "Red"

df["Category"] = df["SRI"].apply(classify)

# Save updated dataset
df.to_csv("students_scored.csv", index=False)

print("Scoring completed and file saved successfully.")
```
